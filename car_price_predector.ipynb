# **Data Cleaning**


1.   From name column get only first three words of name
2.   From years column remove all string value
3.   From Price column replace , with '' and alo remove string values and alo convert it into numeric form
4. From kms_driven remove kms string and alo , and convert into numeric
5. From fuel_type colun remove null



```python
df['name'] = df['name'].str.split(' ').str.slice(0,3).str.join(' ')
```

```python
df = df[df['year'].str.isnumeric()]
df['year'] = pd.to_numeric(df['year'])
```

```python
df = df[df['Price'] != 'Ask For Price']
df['Price'] = df['Price'].str.replace(',','')
df['Price'] = pd.to_numeric(df['Price'])
```

```python
df['kms_driven']=df['kms_driven'].str.split(' ').str[0]
df['kms_driven'] = df['kms_driven'].str.replace(',','')
df = df[df['kms_driven'] != 'Petrol']
df['kms_driven'] = pd.to_numeric(df['kms_driven'])
```

```python
df.dropna(subset=['kms_driven'], inplace=True)
```

```python
df = df[df['fuel_type'].notnull()]
```

```python
df = df.reset_index(drop=True)
```

```python
df.to_csv('clean_data.csv', index=False)
```

# **Building model from here**

```python
x = df.drop(columns='Price')
y = df['Price']
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import  OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
import pickle
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2)
```

```python
oneHotPreprocessing = ColumnTransformer(
    transformers=[
      ('encoder',OneHotEncoder(handle_unknown='ignore'),['name','company','fuel_type'])
    ],
    remainder='passthrough' #keep other columns as they are
)
```

```python
linear_reg = LinearRegression()
```

```python
pipe = Pipeline([
    ('oneHotEncode',oneHotPreprocessing),
    ('linearRegression',linear_reg)
])
```

```python
pipe.fit(x_train,y_train)
```

```python
pickle.dump(pipe,open('carPricePredectionModel.pkl','wb'))
```

```python
pipe.predict(pd.DataFrame([['Maruti Suzuki Swift','Maruti',2019,100,'Petrol']],columns=['name','company','year','kms_driven','fuel_type']))
```
